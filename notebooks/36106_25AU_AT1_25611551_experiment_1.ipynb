### E.4 Model Technical Performance

> Provide some explanations on model performance


```python
# Import performance metrix
from sklearn.metrics import mean_squared_error as mse
from sklearn.metrics import root_mean_squared_error as rmse

#Predict the training data
y_train_pred_T = model_with_intercept.predict(X_train)
y_train_pred_F = model_wthout_intercept.predict(X_train)

#Predict the validation data
y_val_pred_T = model_with_intercept.predict(X_val)
y_val_pred_F = model_wthout_intercept.predict(X_val)

y_test_pred_T = model_with_intercept.predict(X_test)
y_test_pred_F = model_wthout_intercept.predict(X_test)
```

```python
# Calculate RMSE and MSE for training data
train_rmse_T = rmse(y_train, y_train_pred_T)
train_mse_T = mse(y_train, y_train_pred_T)

# Calculate RMSE and MSE for validation data
val_rmse_T = rmse(y_val, y_val_pred_T)
val_mse_T = mse(y_val, y_val_pred_T)

#Calculate RMSE and MSE for testing data
test_rmse_T = rmse(y_test, y_test_pred_T)
test_mse_T = mse(y_test, y_test_pred_T)

# Print the results
print('Train Root Mean Square Error with Intercept: ', train_rmse_T, '\n')
print('Train Mean Square Error with Intercept: ', train_mse_T, '\n')
print('Validation Root Mean Square Error with Intercept: ', val_rmse_T, '\n')
print('Validation Mean Square Error with Intercept: ', val_mse_T, '\n')
print('Test Root Mean Square Error with Intercept: ', test_rmse_T, '\n')
print('Test Mean Square Error with Intercept: ', test_mse_T, '\n')
```

```python
# Calculate RMSE and MSE for training data
train_rmse_F = rmse(y_train, y_train_pred_F)

# Calculate RMSE and MSE for validation data
val_rmse_F = rmse(y_val, y_val_pred_F)

#Calculate RMSE and MSE for testing data
test_rmse_F = rmse(y_test, y_test_pred_F)

# Print the results
print('Train Root Mean Square Error without Intercept: ', train_rmse_F, '\n')
print('Validation Root Mean Square Error without Intercept: ', val_rmse_F, '\n')
print('Test Root Mean Square Error without Intercept: ', test_rmse_F, '\n')
```

```python
import math

# Select numeric and boolean features
X_train_numeric = X_train.select_dtypes(include=['number', 'bool'])
X_val_numeric = X_val.select_dtypes(include=['number', 'bool'])
X_test_numeric = X_test.select_dtypes(include=['number', 'bool'])

# Convert boolean features to 1s and 0s
X_train_numeric = X_train_numeric.astype(float)
X_val_numeric = X_val_numeric.astype(float)
X_test_numeric = X_test_numeric.astype(float)

reg = LinearRegression()

# Assuming 'reg' is your trained LinearRegression model
# Re-fit the model using only numeric features
reg.fit(X_train_numeric, y_train)

# Print the model coefficients
print("Intercept:", model_with_intercept.intercept_)
print("Coefficients:")
for feature, coef in zip(X_train_numeric.columns, reg.coef_[0]):
    print(f"  {feature}: {coef}")

# We use the score method to get r-squared
print('\nR-squared:', reg.score(X_train_numeric, y_train))

# We can also calculate the standard error
stderr = math.sqrt(np.mean((y_train.values.ravel() - reg.predict(X_train_numeric).ravel())**2))
```
